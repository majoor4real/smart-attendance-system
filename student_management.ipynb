```python

#student registration!
students = []
def validate_age ():
  while True:
    try:
      #validate age inputs
      age = int(input("Enter Student age: "))
      if age < 0:
        print("Invalid age. Please enter a positive number.")
      elif age > 120:
        print("Invalid age. Please enter a valid age.")
      elif age == 0:
        print("Invalid age. Please enter a valid age.")
      else:
        return age
    except ValueError:

      print("Invalid input. please enter only numbers.")

def register_student():

  print("Register Student!")

  print("Enter your Details")
  student_id = input("Enter student ID: ")
  name = input("Enter student name: ")

  age = validate_age()

  gender = input("Enter student gender: ")
  department = input("Enter student department: ")
  student = {
    "student_id": student_id,
    "name": name,
    "age": age,
    "gender": gender,
    "department": department
  }
  students.append(student)
  print("\nStudent registered successfully!\n")




#View Student
def view_students():
  if len(students) == 0:
    print("No students found.")
  else:
    print("List of students:")
    for student in students:
      print(f"ID: {student['student_id']}")
      print(f"Name: {student['name']}")
      print(f"Age: {student['age']}")
      print(f"Gender: {student['gender']}")
      print(f"Department: {student['department']}")
      print("-" * 40)


#Search Students
def search_student():
  get_student_id = input("Enter Student Id: ")
  found = False
  for student in students:
    if student["student_id"] == get_student_id:
      print("\nStudent Found:")
      print(f"ID: {student['student_id']}")
      print(f"Name: {student['name']}")
      print(f"Age: {student['age']}")
      print(f"Gender: {student['gender']}")
      print(f"Department: {student['department']}")
      found = True
      break
  if not found:
    print("\nStudent not found.")


#update student
def update_student():
  print("Update Profile!")
  print("========================================")
  print("Enter your Details")
  found = False
  get_student_id = input("Enter Student Id: ")
  for student in students:
    if student["student_id"] == get_student_id:
      student["name"] = input("Enter new name: ")
      student["age"] = int(input("Enter new age: "))
      student["gender"] = input("Enter new gender: ")
      student["department"] = input("Enter new department: ")
      print("\nStudent updated successfully!")
      found = True
      break
  if not found:
    print("\nStudent not found.")


#delete Student
def delete_student():
  print("Delete Profile!")
  print("========================================")
  print("Enter your Details")
  found = False
  get_student_id = input("Enter Student Id: ")
  for student in students:
    if student["student_id"] == get_student_id:
      students.remove(student)
      print("\nStudent deleted successfully!")
      found = True
      break
  if not found:
    print("\nStudent not found.")





#Starting the App!
print("Welcome!")
print("==========Student Management System=============")

while True:
  print("Please SELECT an Option from (1 - 6) ")

  print("1. Add New Student")
  print("2. View Students")
  print("3. Search Student")
  print("4. Edit Student")
  print("5. Delete Student")
  print("6. Quit")

  print("=================================================")
  choice = input("Enter your choice: ")
  if choice == "1":
    print("Please carefully enter your details below!")
    print("=================================================")
    register_student()
  elif choice == "2":
    print("View Your Student")
    view_students()

  elif choice == "3":
    print("Search Student")
    search_student()

  elif choice == "4":
    print("Update Student")
    update_student()
  elif choice == "5":
    print("Delete Student")
    delete_student()
  elif choice == "6":
    print("Thank you for using the Student Management System.")
    break


  else:
    print("Invalid Choice")
```
