```python
import random
def quiz_game():
    user = input("What is your name? ")
    print(f"welcome {user} !")
    quiz = {
    "What is the capital of Nigeria?":{
        "options": [
        "A. Lagos",
        "B. Abuja",
        "C. Kano",
        "D. Ibadan"
      ],
      "answer":"B. Abuja"
      },


      "Which of the following is a programming language?":{
        "options": [
        "A. Python",
        "B. HTML",
        "C. Microsoft Word",
        "D. Photoship"
      ],
      "answer":"A. Python"
      },


      "Which data structure is immutable in Python?":{
        "options": [
        "A. List",
        "B. Set",
        "C. Tuple",
        "D. Dictionary"
      ],
        "answer":"C. Tuple"
      },


      "Which keyword is used to define a function in Python?": {
        "options": [
        "A. function",
        "B. def",
        "C. define",
        "D. func"
      ],
        "answer":"B. def"
      },


      "Which planet is known as the Red Planet?": {
        "options": [
        "A. Earth",
        "B. Jupiter",
        "C. Mars",
        "D. Venus"
      ],
        "answer":"C. Mars"
      }
    }

    count = 1
    scores = 0
    percent =  100/5

    #converting the dict{}. into a list()
    questions = list(quiz.items())

    #shuffling the questoins
    random.shuffle(questions)

    #looping through the shuffled questions
    for question, detail in questions:
      print("------------------")
      print(str(count) + " " + question)
      print("------------------")
      for option in detail["options"]:
        print(option)
      count += 1


      while True:
        u_answer = input("choose from option A - D ").strip().upper()

        if u_answer not in "ABCD":
          print("invalid answer! Please press A, B, C, or D")
          continue

        first_letter = detail["answer"][0].upper()
        if u_answer == first_letter:
          print("correct answer")
          scores += 1
          break
        else:
          print("wrong answer")
          print("the correct answer is " + detail["answer"])
          break


    print()
    print("------------------")
    print("Congratulations!")
    print("------------------")
    print()
    print("you scored " + str(scores) + " points ")
    print(f"Percentage:  {scores * percent} %")
    print()


    if scores == 5:
      print("Your Grade Mark is A")
      print()
    elif scores == 4:
      print("Your Grade Mark is B")
      print()
    elif scores == 3:
      print("Your Grade Mark is C")
      print()
    elif scores == 2:
      print("Your Grade Mark is D")
      print()
    elif scores == 1:
      print("Your Grade Mark is E")
      print()
    else:
      print("Your Grade Mark is F")
      print()


    print(f"you got {scores} out of  {len(quiz)} questions correct.")
    print(f"Thanks for playing {user}")
    play_again = input("Do you want to play again? (Y/N)").strip().upper()
    if play_again == "Y":
      quiz_game()
    else:
      print("Thank you for playing")

print("welcome to my quize")
play_game = int(input("Press 1 to start the game or press 0 to quit!"))
if play_game == 1:
  print("let's play...")
  print()
  quiz_game()
else:
  print("Thank you for playing")
```
